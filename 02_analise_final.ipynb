```python
%sql
-- Média da Tarifa de Uso do Sistema de Distribuição (TUSD) e Tarifa de Energia (TE)
-- por distribuidora, considerando só "Tarifa de Aplicação" (o que o consumidor paga de fato)
SELECT 
    SigAgente,
    SigUF,
    SigRegiao,
    ROUND(AVG(VlrTUSD), 2) AS TUSD_Media,
    ROUND(AVG(VlrTE), 2) AS TE_Media,
    COUNT(*) AS QtdRegistros
FROM mvp_tarifas_energia.gold.fato_tarifas_enriquecida
WHERE DscBaseTarifaria = 'Tarifa de Aplicação'
GROUP BY SigAgente, SigUF, SigRegiao
ORDER BY TUSD_Media DESC
LIMIT 15
```

```python
%sql
SELECT 
    SigAgente,
    SigUF,
    SigRegiao,
    ROUND(AVG(VlrTUSD), 2) AS TUSD_Media,
    ROUND(AVG(VlrTE), 2) AS TE_Media,
    COUNT(*) AS QtdRegistros
FROM mvp_tarifas_energia.gold.fato_tarifas_enriquecida
WHERE DscBaseTarifaria = 'Tarifa de Aplicação'
GROUP BY SigAgente, SigUF, SigRegiao
ORDER BY TUSD_Media ASC
LIMIT 15
```

```python
%sql
SELECT 
    NomClassificacaoAgente,
    COUNT(DISTINCT SigAgente) AS QtdDistribuidoras,
    ROUND(AVG(VlrTUSD), 2) AS TUSD_Media,
    ROUND(AVG(VlrTE), 2) AS TE_Media
FROM mvp_tarifas_energia.gold.fato_tarifas_enriquecida
WHERE DscBaseTarifaria = 'Tarifa de Aplicação'
GROUP BY NomClassificacaoAgente
ORDER BY TUSD_Media DESC
```

```python
%sql
SELECT 
    YEAR(DatInicioVigencia) AS Ano,
    ROUND(AVG(VlrTUSD), 2) AS TUSD_Media,
    ROUND(AVG(VlrTE), 2) AS TE_Media,
    COUNT(*) AS QtdRegistros
FROM mvp_tarifas_energia.gold.fato_tarifas_enriquecida
WHERE DscBaseTarifaria = 'Tarifa de Aplicação'
GROUP BY YEAR(DatInicioVigencia)
ORDER BY Ano
```

```python
%sql
SELECT 
    SigRegiao,
    COUNT(DISTINCT SigAgente) AS QtdDistribuidoras,
    ROUND(AVG(VlrTUSD), 2) AS TUSD_Media,
    ROUND(AVG(VlrTE), 2) AS TE_Media
FROM mvp_tarifas_energia.gold.fato_tarifas_enriquecida
WHERE DscBaseTarifaria = 'Tarifa de Aplicação'
GROUP BY SigRegiao
ORDER BY TUSD_Media DESC
```

```python
%sql
-- Confirma se a média do Sul muda muito ao remover as cooperativas (Permissionárias)
SELECT 
    SigRegiao,
    NomClassificacaoAgente,
    ROUND(AVG(VlrTUSD), 2) AS TUSD_Media,
    COUNT(DISTINCT SigAgente) AS QtdDistribuidoras
FROM mvp_tarifas_energia.gold.fato_tarifas_enriquecida
WHERE DscBaseTarifaria = 'Tarifa de Aplicação' AND SigRegiao IS NOT NULL
GROUP BY SigRegiao, NomClassificacaoAgente
ORDER BY SigRegiao, TUSD_Media DESC
```

```python
# final
%sql
SELECT 
    NomBandeiraAcionada,
    COUNT(*) AS QtdMeses,
    ROUND(AVG(VlrAdicionalAplicadoNoMes), 2) AS ValorMedioAdicional,
    ROUND(MIN(VlrAdicionalAplicadoNoMes), 2) AS ValorMinimo,
    ROUND(MAX(VlrAdicionalAplicadoNoMes), 2) AS ValorMaximo
FROM mvp_tarifas_energia.gold.fato_bandeira_completa
GROUP BY NomBandeiraAcionada
ORDER BY ValorMedioAdicional DESC
```
